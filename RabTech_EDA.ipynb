```python
# Hypothesis 1:
# Higher discount rates are associated with lower customer retention.

from scipy.stats import pearsonr

# Remove missing values
hypothesis1_data = df[["Discount", "Retention"]].dropna()

# Calculate correlation and p-value
correlation, p_value = pearsonr(
    hypothesis1_data["Discount"],
    hypothesis1_data["Retention"]
)

print("Correlation:", round(correlation, 3))
print("P-value:", round(p_value, 4))
```

```python
# Hypothesis 2: Sales and Profit relationship

hypothesis2_data = df[["Sales", "Profit"]].dropna()

correlation2, p_value2 = pearsonr(
    hypothesis2_data["Sales"],
    hypothesis2_data["Profit"]
)

print("Correlation:", round(correlation2, 3))
print("P-value:", round(p_value2, 4))
```

```python
from scipy.stats import f_oneway

# Group profit by product category
profit_groups = [
    group["Profit"].dropna()
    for _, group in df.groupby("Product_Category")
]

# Perform One-Way ANOVA
f_statistic, p_value3 = f_oneway(*profit_groups)

print("F-statistic:", round(f_statistic, 3))
print("P-value:", round(p_value3, 4))
```

```python
print("HYPOTHESIS 1: Discount vs Retention")
print("Correlation:", round(correlation, 3))
print("P-value:", round(p_value, 4))
print("Conclusion:", "Significant relationship" if p_value < 0.05 else "No statistically significant relationship")

print("\nHYPOTHESIS 2: Sales vs Profit")
print("Correlation:", round(correlation2, 3))
print("P-value:", round(p_value2, 4))
print("Conclusion:", "Significant relationship" if p_value2 < 0.05 else "No statistically significant relationship")

print("\nHYPOTHESIS 3: Product Category vs Profit")
print("F-statistic:", round(f_statistic, 3))
print("P-value:", round(p_value3, 4))
print("Conclusion:", "Significant difference between categories" if p_value3 < 0.05 else "No statistically significant difference")
```
